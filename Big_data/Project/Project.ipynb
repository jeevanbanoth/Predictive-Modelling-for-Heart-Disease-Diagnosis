```python
import pandas as pd

def preprocess_file(input_path, output_path):
    try:
        # Read the file with error handling
        df = pd.read_csv(input_path, sep='\t', engine='python', on_bad_lines='skip')

        # Perform any necessary preprocessing steps
        # For example, removing NaN values and resetting the index
        df.dropna(inplace=True)
        df.reset_index(drop=True, inplace=True)

        # Save the preprocessed data to a new file
        df.to_csv(output_path, sep='\t', index=False)
        print(f"Preprocessed file saved to {output_path}")

        # Display a preview of the cleaned data
        print(df.head())
    except pd.errors.ParserError as e:
        print(f"Error parsing the file: {e}")

# Specify input and output file paths
input_test_path = '/Users/jeevankumarbanoth/Downloads/sentiment-analysis-on-movie-reviews/test.tsv'
input_train_path = '/Users/jeevankumarbanoth/Downloads/sentiment-analysis-on-movie-reviews/train.tsv'
output_test_path = '/Users/jeevankumarbanoth/Downloads/sentiment-analysis-on-movie-reviews/preprocessed_test.tsv'
output_train_path = '/Users/jeevankumarbanoth/Downloads/sentiment-analysis-on-movie-reviews/preprocessed_train.tsv'

# Preprocess the test and train files
preprocess_file(input_test_path, output_test_path)
preprocess_file(input_train_path, output_train_path)
```

## Training pre-processed data

```python
# Install textblob library
%pip install textblob

# Install NLTK data
import nltk
nltk.download('punkt')

import pandas as pd
from textblob import TextBlob

# Load the preprocessed datasets
df_test = pd.read_csv('/Users/jeevankumarbanoth/Downloads/sentiment-analysis-on-movie-reviews/preprocessed_test.tsv', sep='\t')
df_train = pd.read_csv('/Users/jeevankumarbanoth/Downloads/sentiment-analysis-on-movie-reviews/preprocessed_train.tsv', sep='\t')

# Function to get sentiment
def get_sentiment(review):
    analysis = TextBlob(review)
    if analysis.sentiment.polarity > 0:
        return 'positive'
    elif analysis.sentiment.polarity == 0:
        return 'neutral'
    else:
        return 'negative'

# Apply the function to the test DataFrame using the 'Phrase' column
df_test['sentiment'] = df_test['Phrase'].apply(get_sentiment)

# Apply the function to the train DataFrame using the 'Phrase' column
df_train['sentiment'] = df_train['Phrase'].apply(get_sentiment)

# Save the results
df_test.to_csv('/Users/jeevankumarbanoth/Downloads/sentiment-analysis-on-movie-reviews/result_test.tsv', sep='\t', index=False)
df_train.to_csv('/Users/jeevankumarbanoth/Downloads/sentiment-analysis-on-movie-reviews/result_train.tsv', sep='\t', index=False)
```

## Counting sentiments

```python
import pandas as pd

# Load the sentiment analysis results
df_test = pd.read_csv('/Users/jeevankumarbanoth/Downloads/sentiment-analysis-on-movie-reviews/result_test.tsv', sep='\t')
df_train = pd.read_csv('/Users/jeevankumarbanoth/Downloads/sentiment-analysis-on-movie-reviews/result_train.tsv', sep='\t')

# Combine the test and train DataFrames
df_combined = pd.concat([df_test, df_train])

# Example: Group by sentiment and count
sentiment_counts = df_combined['sentiment'].value_counts().reset_index()
sentiment_counts.columns = ['sentiment', 'count']

# Save the processed data
sentiment_counts.to_csv('/Users/jeevankumarbanoth/Downloads/sentiment-analysis-on-movie-reviews/Sentiment_Counts.csv', index=False)
```
